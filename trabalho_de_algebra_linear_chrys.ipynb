# ***Vetores com Numpy***

**Soma de Vetores:**

```python
from IPython.core.display import clear_output
import numpy as np

n = int(input("Digite o tamanho dos vetores: "))

va_np = np.zeros(n)
vb_np = np.zeros(n)

print("Vetor A:")
for i in range(n):
   va_np[i] = float(input(f"Digite o {i+1}º número do vetor A: "))

print("Vetor B:")
for i in range(n):
  vb_np[i] = float(input(f"Digite o {i+1}º número do vetor B: "))

vet_soma = va_np + vb_np

clear_output()

print(f"O Vetor A é: {va_np}")
print(f"O Vetor B é: {vb_np}")
print(f"A soma deles é: {vet_soma}")
```

**Subtração de Vetores:**

```python
n = int(input("Digite o tamanho dos vetores: "))

va_np = np.zeros(n)
vb_np = np.zeros(n)

print("Vetor A:")
for i in range(n):
   va_np[i] = float(input(f"Digite o {i+1}º número do vetor A: "))

print("Vetor B:")
for i in range(n):
  vb_np[i] = float(input(f"Digite o {i+1}º número do vetor B: "))

vet_sub = va_np - vb_np

clear_output()

print(f"O Vetor A é: {va_np}")
print(f"O Vetor B é: {vb_np}")
print(f"A subtração deles é: {vet_sub}")
```

**Produto Interno de Vetores:**

```python
n = int(input("Digite o tamanho dos vetores: "))

va_np = np.zeros(n)
vb_np = np.zeros(n)

print("Vetor A:")
for i in range(n):
   va_np[i] = float(input(f"Digite o {i+1}º número do vetor A: "))

print("Vetor B:")
for i in range(n):
  vb_np[i] = float(input(f"Digite o {i+1}º número do vetor B: "))

prod_int = va_np @ vb_np

clear_output()

print(f"O Vetor A é: {va_np}")
print(f"O Vetor B é: {vb_np}")
print(f"O produto interno deles é: {prod_int}")
```

**Norma do Vetor:**

```python
n = int(input("Digite o tamanho do vetor: "))

va_np = np.zeros(n)

print("Vetor:")
for i in range(n):
   va_np[i] = float(input(f"Digite o {i+1}º número do vetor A: "))

norma = np.linalg.norm(va_np)

clear_output()

print(f"O Vetor é: {va_np}")
print(f"A norma dele é: {norma}")
```
